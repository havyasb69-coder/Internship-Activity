```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("C:/SQLite/internship.db")
```

```python
query = """
SELECT interns.name,
       interns.track,
       mentors.mentor_name
FROM interns
INNER JOIN mentors
ON interns.track = mentors.track;
"""
```

```python
df = pd.read_sql_query(query, conn)
```
