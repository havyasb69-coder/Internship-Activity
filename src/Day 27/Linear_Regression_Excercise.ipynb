```python
import  pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("D:/MCA/4TH MCA/DS_AI_Internship/src/Day 27/canada_per_capita_income.csv")

print(df.head())
```

```python
plt.xlabel("year")
plt.ylabel("per capital income (US$)")
plt.scatter(df['year'], df['per capita income (US$)'], color="red", marker="+")
```

```python
year = df.drop("per capita income (US$)",axis = "columns")
year
```

```python
per_capita_income = df['per capita income (US$)']
per_capita_income
```

```python
#equation for linear reg y = mx+c
reg = linear_model.LinearRegression()
reg.fit(year,per_capita_income)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
#to predict the capita income in 2020
new_data = pd.DataFrame([[2020]],columns = ['year'])
prediction = reg.predict(new_data)
prediction = np.round(prediction,2)
prediction
```

```python
#Per capita income of candian in 2020-30 
future_years = pd.DataFrame(range(2020, 2031), columns=['year'])
predictions = reg.predict(future_years)
predictions = np.round(predictions, 2)

result = pd.DataFrame({
    'Year': future_years['year'],
    'Predicted Per Capita Income (US$)': predictions
})

print(result)
```
