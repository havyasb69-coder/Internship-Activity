# Phase 2 – Data Preprocessing (Cleanup)

## Objective
The goal of this phase is to clean the dataset and make it reliable for analysis. This includes identifying missing values, handling them appropriately, and removing duplicate records.

Cleaning data is important because missing or duplicate values can lead to incorrect analysis and misleading results.

```python
df.isnull().sum()
```

```python
(df.isnull().sum() / len(df)) * 100
```

## Missing Value Treatment Decision

Columns with very few missing values were filled using appropriate methods such as mean (for numerical data) or mode (for categorical data).

If any column had a very high percentage of missing values (more than 50%), it would be dropped because it would not provide reliable information.

This ensures the dataset remains accurate while preserving useful information.

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
df.fillna(df.mode().iloc[0], inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

## Duplicate Removal Decision

Duplicate rows were removed to ensure each record represents a unique customer. Keeping duplicates could bias the analysis and produce incorrect insights.

# Phase 3 – Univariate and Bivariate Analysis

## Objective

The goal of this phase is to explore individual variables and relationships between variables using data visualization techniques.

Univariate analysis focuses on understanding the distribution of a single variable, such as Age or AnnualIncome, using histograms and bar charts. This helps identify common values, spread, and possible outliers.

Bivariate analysis examines relationships between two variables, such as AnnualIncome and SpendingScore, using scatter plots and boxplots. This helps identify trends, patterns, and differences between customer groups.


```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
```

```python
plt.figure()
sns.histplot(df['Age'], bins=20, kde=True)
plt.title("Distribution of Customer Age")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

**Explanation:** The histogram shows how customer ages are distributed. It helps identify the most common age group and whether the data is skewed toward younger or older customers.

```python
plt.figure()
sns.histplot(df['AnnualIncome'], bins=20, kde=True)
plt.title("Distribution of Annual Income")
plt.xlabel("Annual Income")
plt.ylabel("Count")
plt.show()
```

**Explanation:** This histogram with KDE curve shows how customer annual income is distributed. It helps identify the most common income ranges and whether the distribution is skewed toward lower or higher incomes.

```python
plt.figure()
sns.countplot(x='Gender', data=df)
plt.title("Number of Customers by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

**Explanation:** This bar chart shows the number of male and female customers. It helps understand the gender composition of the dataset.

```python
plt.figure()
sns.scatterplot(x='AnnualIncome', y='SpendingScore', data=df)
plt.title("Annual Income vs Spending Score")
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.show()
```

**Explanation:** This scatter plot shows the relationship between income and spending score. It helps determine whether higher-income customers tend to spend more.

```python
plt.figure()
sns.boxplot(x='Gender', y='SpendingScore', data=df)
plt.title("Spending Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Spending Score")
plt.show()
```

**Explanation:** This boxplot compares spending behavior between genders. It shows differences in median spending and highlights any outliers.

```python
corr_matrix = df.corr(numeric_only=True)
corr_matrix
```

# Phase 4 – Multivariate Analysis and Storytelling

## Objective

The purpose of this phase is to analyze relationships among multiple variables simultaneously and summarize key findings.

A correlation matrix and heatmap were used to identify strong and weak relationships between numerical variables such as income, spending score, and purchase frequency.

The insights gained from this analysis help understand customer behavior patterns and provide valuable information that can support business decision-making and marketing strategies.


```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix Heatmap")
plt.show()
```

## Correlation Heatmap Explanation

The heatmap shows the relationships between numerical variables such as Age, AnnualIncome, SpendingScore, and PurchaseFrequency. Darker colors indicate stronger relationships. This helps identify which variables influence customer spending behavior the most.
