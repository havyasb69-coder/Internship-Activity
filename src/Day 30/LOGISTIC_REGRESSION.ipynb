```python
 
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

# Dataset
X = np.array([[1],[2],[3],[4],[5],[6]])
y = np.array([0,0,0,1,1,1])

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Train using gradient descent
def train(X, y, lr=0.01, epochs=1000):
    m, n = X.shape
    weights = np.zeros(n)
    bias = 0
    
    for _ in range(epochs):
        linear = np.dot(X, weights) + bias
        y_pred = sigmoid(linear)
        
        dw = (1/m) * np.dot(X.T, (y_pred - y))  # gradient of weights
        db = (1/m) * np.sum(y_pred - y)         # gradient of bias
        
        weights -= lr * dw
        bias -= lr * db
    
    return weights, bias

# Prediction (threshold = 0.5)
def predict_scratch(X, weights, bias):
    linear = np.dot(X, weights) + bias
    y_pred = sigmoid(linear)
    return np.array([1 if i > 0.5 else 0 for i in y_pred])

# Train scratch model
w, b = train(X, y)
y_pred_scratch = predict_scratch(X, w, b)

print("Scratch Predictions:", y_pred_scratch)

# Sklearn model
model = LogisticRegression()
model.fit(X, y)

y_pred_sklearn = model.predict(X)
y_prob = model.predict_proba(X)

print("\nSklearn Predictions:", y_pred_sklearn)
print("Probabilities:\n", y_prob)

# Evaluation
print("\nAccuracy:", accuracy_score(y, y_pred_sklearn))
print("Confusion Matrix:\n", confusion_matrix(y, y_pred_sklearn))

# Visualization
X_test = np.linspace(0, 7, 100).reshape(-1,1)
y_prob_curve = model.predict_proba(X_test)[:,1]

plt.scatter(X, y)
plt.plot(X_test, y_prob_curve)
plt.title("Logistic Regression Curve")
plt.show()

# Test new value
test_value = np.array([[3.5]])
print("\nNew Prediction:", model.predict(test_value))
print("Probability:", model.predict_proba(test_value))
```
